# Atividade 2 - Algoritmos e Métodos Numéricos, Erros, Medidas e Fontes

Uma equipe de desenvolvedores e engenheiros está projetando um sistema de monitoramento térmico em tempo real para as CPUs de um Data Center. O objetivo do sistema é acionar o resfriamento de emergência sempre que a temperatura da CPU atingir o limite crítico de segurança, definido como $85\text{ }^\circ\text{C}$ por um sensor óptico padrão de calibração de altíssima precisão.
Para baratear o custo, a equipe testou 6 módulos de sensores de temperatura IoT diferentes (série de nós de sensores $S_1$ a $S_6$). Devido a limitações de hardware — como a resolução do Conversor Analógico-Digital (ADC) do microcontrolador (erro de quantização e truncamento), ruído eletromagnético e tolerância dos componentes —, os valores digitais lidos por cada sensor apresentaram pequenas divergências em relação ao valor real de referência.



Os resultados dos sensores estão na tabela abaixo:

| Módulo de Sensor | Temperatura Lida (°C) |
| --- | --- |
| Sensor S1 | 84,820 |
| Sensor S2 | 85,150 |
| Sensor S3 | 84,985 |
| Sensor S4 | 85,040 |
| Sensor S5 | 85,310 |
| Sensor S6 | 84,910 |


Seu objetivo é analisar esses dados e calcular diferentes métricas de erro, usando as bibliotecas pandas e numpy.

---


```python
from IPython.display import display
# Importando bibliotecas

#Analise de dados
import pandas as pd
import numpy as np
```

```python
Valor_referencia = 85
#Tabela de dados
tabela = {
    "Sensor": [
        "Sensor S1",
        "Sensor S2",
        "Sensor S3",
        "Sensor S4",
        "Sensor S5",
        "Sensor S6",
    ],
    "Temp_Lida (°C)": [84.820, 85.150, 84.985, 85.040, 85.310, 84.910],
}

dados = pd.DataFrame(tabela)
display(dados)
```

### Parte 1 - Erro Absoluto, Erro Relativo e Erro Percentual

Calcule o Erro Absoluto, o Erro Relativo e o Erro Percentual dos valores medidos.

```python
dados['Erro_Absoluto'] = abs(Valor_referencia - dados['Temp_Lida (°C)'])
display(dados)
```

```python
dados['Erro_Relativo'] = dados['Erro_Absoluto'] / abs(dados['Temp_Lida (°C)'])
display(dados)
```

```python
dados['Erro_Percentual'] = dados['Erro_Relativo'] * 100
display(dados)
```

**Formulário para cálculo de erros**

- Erro Absoluto:  
$$E_a = |V_t - V_m|$$

- Erro Relativo:  
$$E_r = \frac{E_a}{|V_m|}$$

- Erro Percentual:  
$$E_p = E_r \times 100\%$$

Onde:  
- $V_t$ = valor teórico de referência  
- $V_m$ = valor medido pelo sensor

### Parte 2 - Arredondamento e Truncamento

Um engenheiro biomédico está testando um novo sensor para um oxímetro de pulso hospitalar (aparelho que mede a saturação de oxigênio no sangue, $SpO_2$).

O sensor faz a medição com alta precisão (quatro casas decimais), mas o visor do aparelho só tem espaço para mostrar o número arredondado com duas casas decimais.

As medições feitas pelo sensor em cinco pacientes foram:

| Paciente | Oxigênio Medido (%) |
| --- | --- |
| P1 | 92.3456 |
| P2 | 98.8765 |
| P3 | 95.5555 |
| P4 | 97.4999 |
| P5 | 90.0501 |

### 1 - Fazer um DataFrame para organização visual dos dados

```python
import pandas as pd

# Dados dos pacientes
dados = {
    "Paciente": ["P1", "P2", "P3", "P4", "P5"],
    "Oxigênio Medido (%)": [92.3456, 98.8765, 95.5555, 97.4999, 90.0501]
}

# Criando o DataFrame
df = pd.DataFrame(dados)

# Exibindo o DataFrame
display(df)
```

### 2 - Arredonde todos os valores para 2 casas decimais

```python
import pandas as pd

# Criando o DataFrame
dados = {
    "Paciente": ["P1", "P2", "P3", "P4", "P5"],
    "Oxigênio Medido (%)": [92.3456, 98.8765, 95.5555, 97.4999, 90.0501]
}

df = pd.DataFrame(dados)

# Arredondando os valores numéricos para 2 casas decimais
df["Oxigênio Medido (%)"] = df["Oxigênio Medido (%)"].round(2)

# Exibindo o DataFrame
display(df)
```

### 3 - Trunque os valores para 1 casa decimal

```python
import pandas as pd
import numpy as np

# Criando o DataFrame
dados = {
    "Paciente": ["P1", "P2", "P3", "P4", "P5"],
    "Oxigênio Medido (%)": [92.3456, 98.8765, 95.5555, 97.4999, 90.0501]
}

df = pd.DataFrame(dados)

# Truncando para 1 casa decimal
df["Oxigênio Medido (%)"] = np.trunc(df["Oxigênio Medido (%)"] * 10) / 10

# Exibindo o DataFrame
display(df)
```

### 4 - Qual é o valor inteiro mais próximo para cada?

```python
import pandas as pd
import numpy as np

# Criando o DataFrame
dados = {
    "Paciente": ["P1", "P2", "P3", "P4", "P5"],
    "Oxigênio Medido (%)": [92.3456, 98.8765, 95.5555, 97.4999, 90.0501]
}

df = pd.DataFrame(dados)

# Arredondando para o inteiro mais próximo
df["Inteiro Mais Próximo"] = np.round(df["Oxigênio Medido (%)"]).astype(int)

display(df)
```

### 5 - Mostre os resultados dentro de uma tabela

```python
import pandas as pd
import numpy as np
from IPython.display import display

# Criando o DataFrame
dados = {
    "Paciente": ["P1", "P2", "P3", "P4", "P5"],
    "Oxigênio Medido (%)": [92.3456, 98.8765, 95.5555, 97.4999, 90.0501]
}

df = pd.DataFrame(dados)

# Arredondar para 2 casas
df["Arredondado (2 casas)"] = np.round(df["Oxigênio Medido (%)"], 2)

# Truncar para 1 casa
df["Truncado (1 casa)"] = np.trunc(df["Oxigênio Medido (%)"] * 10) / 10

# Inteiro mais próximo
df["Inteiro mais próximo"] = np.round(df["Oxigênio Medido (%)"]).astype(int)

# Exibir tabela
display(df)
```
